# Recency-Frequency-Monetary (RFM) analysis

```python
# defining the fatures to be part of the RFM analysis

rfm_features = ['CustomerID', 'MonthsInService', 'MonthlyRevenue', 'MonthlyMinutes',
                'TotalRecurringCharge', 'Churn']

# summary statistics for the features
train_data1[rfm_features].describe()
```

```python
# building the rfm features

rfm = train_data1[['CustomerID', 'MonthsInService', 'MonthlyRevenue', 'MonthlyMinutes', 'Churn']].copy()

rfm['Recency'] = rfm['MonthsInService'].max() - rfm['MonthsInService'] 
rfm['Frequency'] = rfm['MonthlyMinutes']
rfm['Monetary'] = rfm['MonthlyRevenue']

rfm.head()
```

```python
# scoring each dimension into quintiles

rfm['R_score'] = pd.qcut(rfm['Recency'].rank(method='first'), q=5, labels=[1, 2, 3, 4, 5])
rfm['F_score'] = pd.qcut(rfm['Frequency'].rank(method='first'), q=5, labels=[1, 2, 3, 4, 5])
rfm['M_score'] = pd.qcut(rfm['Monetary'].rank(method='first'), q=5, labels=[1, 2, 3, 4, 5])

# making sure its int
rfm[['R_score', 'F_score', 'M_score']] = rfm[['R_score', 'F_score', 'M_score']].astype(int)

# adding the features to form rfm_score
rfm['RFM_score'] = rfm['R_score'] + rfm['F_score'] + rfm['M_score']

# summary statistics of the features
rfm[['R_score', 'F_score', 'M_score', 'RFM_score']].describe()
```

```python
sns.heatmap(data=rfm[['R_score', 'F_score', 'M_score', 'RFM_score']].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation for rfm features')
plt.tight_layout()
plt.show()
```

There's a noticeably high correlation between F_score and FRM_score, M_scor and RFM_score, as well as M_score and F_score..

```python
# creating segments and assigning them
def telecom_segment(row):
    r, f, m = row['R_score'], row['F_score'], row['M_score']
    if r >= 4 and f >= 4 and m >= 4:
        return 'CHAMPIONS'
    elif r >= 3 and f >= 3:
        return 'LOYAL'
    elif r >= 4 and f <= 2:
        return 'NEW / LIGHT'
    elif r <= 2 and f >= 4:
        return 'AT RISK'
    elif r <=2 and m >= 4:
        return 'HIGH VALUE AT RISK'
    elif r <=2 and f <= 2:
        return 'LOST'
    else:
        return 'MID-TIER'

rfm['Segment'] = rfm.apply(telecom_segment, axis=1)

# checking value counts
rfm['Segment'].value_counts()
```

From dataframe above we can see that our data contains 'LOYAL' customers as majority with 13181 customers. These are the normal everyday customers that are consistently active. 'LOST' customers are those that have already atrittioned, they have gone dark. Then we those that are 'AT RISK', probably thinking about moving somewhere else. Our 'CHAMPIONS', with a total number of 6392 customers, these are our premium customers, who spend and engage heavily. There are also those customers who provide so much have but are at risk of churning, 'HIGH VALUE AT RISK', the minority group.

# MODEL DEVELOPMENT

```python
#Import libraries for the machine learning pipeline
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder, LabelEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV, StratifiedKFold, cross_val_score
from imblearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.decomposition import PCA
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn import svm
from sklearn.metrics import f1_score, roc_auc_score, classification_report, confusion_matrix
import catboost as cb
from imblearn.over_sampling import SMOTE
from imblearn.combine import SMOTETomek
from sklearn.neighbors import KNeighborsClassifier
from sklearn.feature_selection import RFE
```

```python
# Splitting the dataset
X = train_data1.drop(columns=['CustomerID', 'Churn', 'ServiceArea', 'ServiceArea_grouped'])
y = train_data1['Churn']

X_train, X_val, y_train, y_val = train_test_split(X, y, stratify=y, test_size=0.2, random_state=42)

y_train = y_train.astype(int)
y_val = y_val.astype(int)
```

```python
print(f'{categorical_features}\n')

print(numerical_features)
```

# Outlier handling with winsorization

```python
# The winsorization technique will be applied to both X_train and X_test separately to avoid data leakage
for col in numerical_features:
    Q1 = X_train[col].quantile(0.25)
    Q3 = X_train[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    X_train[col] = X_train[col].clip(lower_bound, upper_bound)
    X_val[col] = X_val[col].clip(lower_bound, upper_bound)
```

# Models, parameters and preprocessor 

```python
# Defining models

# creating scale_pos_weight for xgboost
neg = (y_train == 0).sum()
pos = (y_train == 1).sum()
scale_pos_weight = neg / pos

models = {
    'XGBoost': XGBClassifier(scale_pos_weight=scale_pos_weight, random_state=42, eval_metric='logloss'),
    'LightGBM': LGBMClassifier(class_weight='balanced', random_state=42, verbose=-1),
    'CatBoost': cb.CatBoostClassifier(auto_class_weights='Balanced', random_state=42, verbose=0)
}

# defining parameters
xgb_grid = {
    'classifier__n_estimators': [200,400,600],
    'classifier__learning_rate': [0.01,0.05,0.1],
    'classifier__max_depth': [3,5,7],
    'classifier__min_child_weight': [1,3,5],
    'classifier__subsample': [0.7,0.8,1],
    'classifier__colsample_bytree': [0.7,0.8,1],
    'classifier__gamma': [0,0.1,0.3]
}

lgbm_grid = {
    'classifier__n_estimators': [200,400,600],
    'classifier__learning_rate': [0.01,0.05,0.1],
    'classifier__max_depth': [5,10,15],
    'classifier__num_leaves': [31,50,100],
    'classifier__subsample': [0.7,0.8,1],
    'classifier__colsample_bytree': [0.7,0.8,1]
}

cat_grid = {
    "classifier__depth": [4, 6, 8],
    "classifier__learning_rate": [0.01, 0.03, 0.1],
    "classifier__iterations": [200, 500, 1000],
    "classifier__l2_leaf_reg": [1, 3, 5, 7],
    "classifier__border_count": [32, 64, 128]
}

params_grid = {
    'XGBoost': xgb_grid,
    'LightGBM': lgbm_grid,
    'CatBoost': cat_grid
}

# defining the prerocessors

preprocessor = ColumnTransformer([
    ('cat', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'), categorical_features),
    ('num', StandardScaler(), numerical_features)
])

# Using stratifies K-fold for class imbalance
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

# Baseline models

```python
# Looping through the models
for name, model in models.items():
    print(f'Training {name} model.....')

    # defining the pipeline
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])

    # training the models
    pipeline.fit(X_train, y_train)

    # predictions
    y_predict = pipeline.predict(X_val)
    y_prob = pipeline.predict_proba(X_val)[:, 1]
    # Model evaluations
    print(f'ROC-AUC score: {roc_auc_score(y_val, y_prob)}')
    print(f'Classification Report: \n{classification_report(y_val, y_predict)}')
    print(f'Confusion Matrix: \n{confusion_matrix(y_val, y_predict)}')
    cv_scores = cross_val_score(
        pipeline,
        X_train,
        y_train,
        cv=kf,
        scoring='roc_auc'
    )
    print(f'Cross Validation (roc auc): \n {cv_scores.mean():.4f} ± {cv_scores.std():.4f}')
    print('*' * 53)
```

Though the baseline models are performing very well on negative instances, they are performing poorly on positive instances

# Hyperparameter tuning with randomized_search

```python
# Looping through the models
best_models = {}
for name, model in models.items():
    print(f'Training {name} model.....')

    # defining the pipeline
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])

    # defining the hyperparameter tuning method
    random_search = RandomizedSearchCV(
        pipeline,
        params_grid[name],
        n_iter=50,
        cv=kf,
        n_jobs=-1,
        scoring='roc_auc'
    )

    # training the models
    random_search.fit(X_train, y_train)

    # best models and making predictions
    best_models[name] = random_search.best_estimator_
    y_predict = best_models[name].predict(X_val)
    y_prob = best_models[name].predict_proba(X_val)[:, 1]
    
    # Model evaluations
    print(f'{name} Best params: {random_search.best_params_}')
    print(f'ROC-AUC score: {roc_auc_score(y_val, y_prob)}')
    print(f'Classification Report: \n{classification_report(y_val, y_predict)}')
    print(f'Confusion Matrix: \n{confusion_matrix(y_val, y_predict)}')
    print('*' * 53)
```

The gradient boosting models seem to be having a similar level of performance, with 40% precision, 65% recall and 50% f1 score, and because of that, I'll go ahead and use CatBoost as my selected model.

```python
# determing the optimal threshold
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = np.arange(0.1, 0.91, 0.05)

results = []

for threshold in thresholds:

    y_pred = (y_prob >= threshold).astype(int)

    precision = precision_score(y_val, y_pred)
    recall = recall_score(y_val, y_pred)
    f1 = f1_score(y_val, y_pred)

    results.append({
        'Threshold': threshold,
        'Precision': precision,
        'Recall': recall,
        'F1': f1
    })

threshold_df = pd.DataFrame(results)

threshold_df
```

The optimal threshold is around 0.4 with a precision score of 37% and a recall of 85% and an f1 score of 49&.
The idea is for us to be able to catch more churners, though it would mean that with a decrease in precision means, more false positives, some we will live with.
